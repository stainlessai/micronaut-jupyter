```groovy
// The cell's value (last expression) is recorded in the notebook as an
// execute_result, unlike println output which is redirected to the server log.
def outcome
try {
    Thread.sleep(30000)  // 30 seconds, interrupted if the kernel is killed
    outcome = "SLEEP_RESULT: SUCCESS after " + (System.currentTimeMillis() - startTime) + "ms"
} catch (Throwable t) {
    outcome = "SLEEP_RESULT: INTERRUPTED by " + t.class.name + " after " + (System.currentTimeMillis() - startTime) + "ms"
}
outcome
```
