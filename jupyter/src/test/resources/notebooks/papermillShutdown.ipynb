```groovy
import io.micronaut.http.client.HttpClient
import io.micronaut.http.HttpRequest
import java.net.URL

println "Cell 1: Starting papermill shutdown test with active Netty threads"

// Create HTTP clients that spawn Netty event loop threads (same as production)
def clients = []
try {
    for (int i = 0; i < 5; i++) {
        def client = HttpClient.create(new URL("http://localhost:8080"))
        clients.add(client)
    }
    println "Created ${clients.size()} HTTP clients with active Netty event loops"
} catch (Exception e) {
    println "Note: Could not create HTTP clients: ${e.message}"
}
```

```groovy
// Simulate a long batch job: many items, each emitting output
// (iopub traffic the client must drain at shutdown)
def items = (1..951).collect { it }
items.each { item ->
    println "Processing item ${item} of ${items.size()}"
}
println "Processed ${items.size()} items successfully"
```

```groovy
// Leave HTTP clients open (don't close them) - simulates production where
// Netty event loop threads from HTTP clients are still alive at shutdown time
println "Cell 3: All work complete, notebook done"
println "Active threads: ${Thread.activeCount()}"
Thread.getAllStackTraces().keySet().findAll { it.name.contains('nioEventLoop') || it.name.contains('other-') }.each {
    println "  Active Netty thread: ${it.name}"
}
```
